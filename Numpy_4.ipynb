```python
logo = np.load('numpy logo.npy')
plt.figure(figsize=(10 , 5))
plt.subplot(121) # this means 1 row, 2 columns, and this is the first plot
plt.imshow(logo)
plt.title("Numpy Logo")

dark_logo = 1 - logo
plt.figure(figsize=(10 , 5))
plt.subplot(121) # this means 1 row, 2 columns, and this is the second plot
plt.imshow(dark_logo)
plt.title("Dark Numpy Logo")
plt.xlabel('Range -->')
plt.ylabel('super_range -->')
```
