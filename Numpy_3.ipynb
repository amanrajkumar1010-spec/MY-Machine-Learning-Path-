### ON A SMALL DATA OF RESTURENT using Numpy and Matplotlib(mini use) 


```python
import numpy as np
import matplotlib.pyplot as plt  # used for graph plotting

# Restaurant sales data from 2020–2025
# Columns: [Restaurant_ID, Sales_2020, Sales_2021, Sales_2022, Sales_2023, Sales_2024, Sales_2025]

restaurant_sales = np.array([
    [101, 120000, 135000, 150000, 165000, 180000, 195000], # Baba Ganoush
    [102, 95000,  110000, 125000, 140000, 155000, 170000], # Tandoori Delight
    [103, 143000, 150000, 162000, 175000, 188000, 200000], # La Piazza
    [104, 87000,  92000,  98000,  105000, 112000, 120000], # Bistro Paris
    [105, 160000, 172000, 185000, 198000, 210000, 225000], # Sushi Master
    [106, 130000, 145000, 158000, 170000, 185000, 199000] # The Green Leaf
    ])

print("Restaurant Sales Data:\n")
print(restaurant_sales)

# now adding data of 2026
new_sales_2026 = np.array([[210000], [185000], [220000], [130000], [240000], [210000]])
restaurant_sales = np.hstack((restaurant_sales, new_sales_2026))
print("\nUpdated Restaurant Sales Data with 2026:\n")
print(restaurant_sales)


# adding data of one more restaurant
new_restaurant_sales = np.array([[107, 110000, 125000, 140000, 155000, 170000, 185000, 200000]])
restaurant_sales = np.vstack((restaurant_sales, new_restaurant_sales))
print("\nUpdated Restaurant Sales Data with New Restaurant:\n")
print(restaurant_sales)
```

```python

# sales of first 3 restaurants from 2020 to 2026
print("sample data of 1st 3 resturent:\n",restaurant_sales[:3])
print("all data without restaurant id:\n",restaurant_sales[:,1:])
print("total sale per year regardless of resturent:\n",np.sum(restaurant_sales[:,1:],axis=0)) # column wise sum
print("total sale per year regardless of resturent:\n",np.array(np.sum(restaurant_sales[:,1:],axis=0))) # column wise sum

print(sum(map(lambda x: int(x),[120000, 135000, 150000, 165000, 180000, 195000, 210000]))) # checking sum of 1st restaurant sales from 2020 to 2026
print(np.sum(np.array([120000,135000,150000, 165000, 180000 ,195000, 210000]))) # another way to check sum of 1st restaurant sales from 2020 to 2026
print(np.sum(np.array(restaurant_sales[0,1:]))) # another way to check sum of 1st restaurant sales from 2020 to 2026

print("total sale per restaurant regardless of year:\n",np.array(np.sum(restaurant_sales[:,1:],axis=1))) # row wise sum
print("total sale for 1st and 2nd restaurants:\n",np.array(np.add(restaurant_sales[:,1],restaurant_sales[:,2])))
```

```python
# minimum sale per restaurant across all years
print("minimum sale per year across all restaurants:\n",np.min(restaurant_sales[:,1:],axis=0)) # column wise min
print("minimum sale per restaurant across all years:\n",np.min(restaurant_sales[:,1:],axis=1)) # row wise min

# maximum sale per restaurant across all years
print("maximum sale per restaurant across all years:\n",np.max(restaurant_sales[:,1:],axis=1)) # column wise max
print("maximum sale per year across all restaurants:\n",np.max(restaurant_sales[:,1:],axis=0)) # row wise max

# sales of resturant in all years in ascending order
print (np.sort(np.sum(restaurant_sales[:,1:],axis=1)))

# short out the id of resturent which have sale less than 1 million in total
total_sales_per_restaurant = np.sum(restaurant_sales[:,1:],axis=1)
low_sales_restaurants = np.where(total_sales_per_restaurant < 1000000, restaurant_sales[:,0], 0) #condition on each element of total_sales_per_restaurant, if condition is true then return restaurant id else return 0
print("IDs of restaurants with total sales less than 1 million:\n", low_sales_restaurants)

# AVERAGE SALE PER RESTAURANT ACROSS ALL YEARS
average_sales_per_restaurant = np.mean(restaurant_sales[:,1:], axis=1)
print("Average sales per restaurant across all years:\n", average_sales_per_restaurant)

# AVERAGE SALE PER RESTAURANT ACROSS ALL YEARS
average_sales_per_year = np.sum(restaurant_sales[:,1:], axis=1) / restaurant_sales.shape[0] #works as scalar division as we are dividing by number of restaurants , resturent.shape[0] gives number of rows in restaurant_sales array as shape gives tuple of shape(num. of row,num. of column) and [0] gives its 1st element
print("Average sales per year across all restaurants:\n", average_sales_per_year)
print(restaurant_sales.shape)

# cumulative sales per restaurant across all years
cumulative_sales_per_restaurant = np.cumsum(restaurant_sales[:,1:], axis=1) # yearly increassing sales for each restaurant row wise 
print("Cumulative sales per restaurant across all years:\n", cumulative_sales_per_restaurant)

```

```python
# Create an array of years from 2020 to 2026
years = np.arange(2020, 2027) #[2020 2021 2022 2023 2024 2025 2026]

# Create figure with better aspect ratio
plt.figure(figsize=(12, 8))

# Loop through each restaurant
for i in range(restaurant_sales.shape[0]):

    # Plot sales trend for each restaurant
    plt.plot(
        years, # x-axis
        restaurant_sales[i, 1:], # Y-axis
        marker='p',          # cleaner circular markers
        linewidth=1,         # thicker lines
        markersize=8,        # marker size
        label=f'Restaurant {int(restaurant_sales[i, 0])}' #side Bar
    )

# Add title with styling
plt.title('Restaurant Sales from 2020 to 2026',fontsize=16, fontweight='bold')

# Axis labels
plt.xlabel('Year', fontsize=12)
plt.ylabel('Sales', fontsize=12)

# Show all years on x-axis
plt.xticks(years)

# Add light grid
plt.grid(True, linestyle='--', alpha=0.6)

# Place legend outside plot for better readability
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

# Automatically adjust layout
plt.tight_layout()

# Display plot
plt.show()
```

## Operation on vectors

```python
vector_1 = np.array([2,5,8])
vector_2 = np.array([1,3,4])
print("vector sum:", np.add(vector_1, vector_2))
print("vector difference:", np.subtract(vector_1, vector_2))
print("vector sum2:", vector_1 + vector_2)
print("vector product:", vector_1 * vector_2)
#  actual vector product:
print("actual vector product:", np.dot(vector_1, vector_2))

# now Matrix multiplication
matrix_1 = np.array([[1,2,3],[3,4,8],[2,4,6]])
matrix_2 = np.array([[5,6,8],[7,8,9],[1,2,3 ]])
print("matrix product:\n", np.dot(matrix_1, matrix_2))
print("matrix product using @ operator:\n", matrix_1 @ matrix_2) 
print("matrix product using np.matmul:\n", np.matmul(matrix_1, matrix_2))


# finding angle between two vectors
def angle_between_vectors(v1, v2):
    # Calculate the dot product
    dot_product = np.dot(v1, v2)
    
    # Calculate the magnitudes of the vectors
    magnitude_v1 = np.linalg.norm(v1) #linear Algebra normal(Magnitude) of V1
    magnitude_v2 = np.linalg.norm(v2) ##linear Algebra normal(Magnitude) of V2
    
    # Calculate the cosine of the angle
    cos_angle = dot_product / (magnitude_v1 * magnitude_v2)
    print("cosine of angle:", cos_angle)
    
    # Ensure the cosine value is within the valid range [-1, 1]
    # cos_angle = np.clip(cos_angle, -1.0, 1.0)
    """3. Cosine ko valid range mein rakhna
# cos_angle = np.clip(cos_angle, -1.0, 1.0)

Ye line abhi commented hai, kyunki # laga hua hai.

Cosine ki mathematical range hamesha:
$$ -1 \leq \cos(\theta) \leq 1 $$
honi chahiye.
np.clip() value ko -1 aur 1 ke beech force karta hai.

Example:
np.clip(1.2, -1, 1)
result:
1.0
Aur:
np.clip(-1.5, -1, 1)
result:
-1.0"""
    
    # Calculate the angle in radians and then convert to degrees
    angle_rad = np.arccos(cos_angle)  #inverse of cos use to convert the value of cos to radian 
    angle_deg = np.degrees(angle_rad)
    return angle_deg
vector_a = np.array([2, 5, 8])
vector_b = np.array([1, 3, 4])
angle = angle_between_vectors(vector_a, vector_b)
print(f"The angle between the vectors is: {angle:.2f} degrees")

# Vectors → cos(θ) value → arccos() → radians → np.degrees() → degrees.

```

| Code               | Meaning                                   |
| ------------------ | ----------------------------------------- |
| `dot_product`      | Vectors ka dot product                    |
| `magnitude_v1`     | First vector ki length                    |
| `magnitude_v2`     | Second vector ki length                   |
| `cos_angle`        | Angle ka cosine                           |
| `np.clip()`        | Cosine ko `-1` se `1` ke beech rakhta hai |
| `np.arccos()`      | Cosine se angle nikalta hai               |
| `angle_rad`        | Angle in radians                          |
| `np.degrees()`     | Radians → degrees                         |
| `angle_deg`        | Final angle in degrees                    |
| `return angle_deg` | Final answer return karta hai             |
