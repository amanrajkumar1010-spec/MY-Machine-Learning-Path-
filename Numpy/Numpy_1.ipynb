### Efficiency checking of array and list


```python
import time
import numpy as np


# ============================================================
# 1. PYTHON LIST
# ============================================================

start = time.time()                         # Start the timer

x = list(i for i in range(1000000))         # Create a Python list
                                             # containing 1,000,000 numbers

end = time.time()                           # Stop the timer

print("List Time taken:", end - start)      # Print execution time


# ============================================================
# 2. PYTHON LIST → NUMPY ARRAY
# ============================================================

start = time.time()                         # Start the timer

x = np.array([i for i in range(1000000)])   # First create a Python list
                                             # Then convert it into a NumPy array

end = time.time()                           # Stop the timer

print("NumPy Array Time taken:", end - start)


# ============================================================
# 3. NUMPY arange()
# ============================================================

start = time.time()                         # Start the timer

p = np.arange(1000000)                      # Directly create a NumPy array
                                             # containing numbers 0 to 999999

end = time.time()                           # Stop the timer

print("NumPy Time taken:", end - start)


# ============================================================
# 4. PYTHON range()
# ============================================================

start = time.time()                         # Start the timer

p = range(1000000)                          # Create a Python range object
                                             # Numbers are generated when needed

end = time.time()                           # Stop the timer

print("Python range Time taken:", end - start)
```

### Conclusion

* `range()` creates a Python range object and generates values when needed, so it is very fast and memory efficient.
* `list()` stores all the elements in a Python list, so it requires more memory.
* `np.array()` converts a Python list into a NumPy array, so it involves an extra conversion step.
* `np.arange()` directly creates a NumPy array and is generally more efficient for creating numerical sequences.
* Therefore, **`range()` and `np.arange()` are not exactly equivalent**: `range()` creates a sequence object, while `np.arange()` creates and stores an actual NumPy array.
* For numerical computing and large arrays, **NumPy is generally preferred** because it provides efficient storage and fast numerical operations.

## Array reshaping

```python
import numpy as np

# Create array with values 0 to 19
arr = np.arange(20)
print("Original:", arr)

# Access first element
print("arr[0]:", arr[0])


# Reshape: changes shape, returns a new array
reshaped_arr = arr.reshape((5, 4))
print("Reshaped:\n", reshaped_arr)


# 3D reshape: 4 × 5 × 1 = 20 elements
reshaped_arr_3d = arr.reshape((4, 5, 1))
print("3D Reshaped:\n", reshaped_arr_3d)


# Resize: changes the original array in-place
arr.resize((4, 5))
print("Resized:\n", arr)


# Flatten: converts to 1D and returns a COPY
flattened = reshaped_arr.flatten()
print("Flattened:", flattened)


# Ravel: converts to 1D and returns a VIEW when possible
ravelled = reshaped_arr.ravel()
print("Ravelled:", ravelled)


# Transpose: swaps rows and columns
transposed = reshaped_arr.T
print("Transposed:\n", transposed)


# Adjoint: conjugate transpose
# For real numbers, adjoint = transpose
adjoint = reshaped_arr.conj().T
print("Adjoint:\n", adjoint)
```

             → new shaped array return karta hai

resize()  → original array ko HI change karta hai
             → in-place modify karta hai
