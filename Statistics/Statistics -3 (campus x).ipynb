# PMF to CDF

```python
# REVISION: Simulate two dice 10,000 times and store their sums.
import pandas as pd  # Pandas: data handling / Series, DataFrame
import random  # Random-number generation
L = []  # Empty list to store sums of two dice
for i in range(10000):  # Repeat the experiment 10,000 times
  a = random.randint(1,6) # For a Random Dice roll for 10000 times
  b = random.randint(1,6)  # Random value from 1 to 6 for die 2

  L.append(a + b) # addind both dice 
  print(L)
```

```python
# REVISION: Count how many times each sum occurs.
len(L)  # Number of simulated observations
pd.Series(L).value_counts()  # Frequency of each possible sum
```

```python
# REVISION: Frequency / total frequency gives the empirical PMF.
s = (pd.Series(L).value_counts()/pd.Series(L).value_counts().sum()).sort_index()  # Relative frequency = estimated PMF; sort by outcome
s  # Display the estimated PMF
```

```python
# REVISION: Cumulative PMF gives the CDF.
import numpy as np  # Numerical computing / cumulative operations
np.cumsum(s)  # Cumulative sum of PMF values = CDF
```

```python
# REVISION: Visualize PMF.
s.plot(kind='bar')  # Bar plot of PMF
```

```python
# REVISION: Visualize CDF.
np.cumsum(s).plot(kind='bar')  # Bar plot of CDF
```

# Parametric Density Estimation

```python
# REVISION: Create a normal sample for parametric density estimation.
import numpy as np  # Numerical arrays and linspace
import matplotlib.pyplot as plt  # Plotting
import seaborn as sns  # Statistical visualization
from numpy.random import normal  # Generate normally distributed random data

sample = normal(loc=50,scale=5,size=1000)  # Mean=50, SD=5, generate 1000 observations
```

```python
# REVISION: First inspect the distribution with a histogram.
# plot histogram to understand the distribution of Data
plt.hist(sample,bins=10)  # Histogram shows the shape of the sample distribution
```

```python
# REVISION: Estimate the two parameters: mean and standard deviation.
# calculate the sample mean and sample std
sample_mean = sample.mean()  # Estimated mean of the sample
sample_std = sample.std()  # Estimated standard deviation of the sample
```

```python
# REVISION: Build a normal distribution using those estimated parameters.
# fit the distribution with the above parameters
from scipy.stats import norm  # Normal distribution object
dist = norm(sample_mean,sample_std)  # Fit a normal distribution using estimated mean and SD
```

```python
# REVISION: Evaluate the fitted PDF at selected x-values.
values = np.linspace(sample.min(),sample.max(),100)  # 100 x-values across the observed range
probabilities = [dist.pdf(value) for value in values]  # PDF gives density at each x-value
```

```python
pd.DataFrame(values,probabilities).head(5)  # View first 5 value-density pairs (index/value arrangement)
```

```python
# REVISION: Compare the normalized histogram with the fitted PDF.
plt.hist(sample,bins=10,density=True)  # Normalize histogram so it represents density
plt.plot(values,probabilities)  # Overlay fitted normal PDF curve
```

```python
# REVISION: Seaborn gives a convenient distribution visualization.
import seaborn as sns  # Seaborn statistical plotting
sns.distplot(sample)  # Histogram + density-style visualization
```

# KDE

```python
# REVISION: Create a mixture of two normal distributions.
sample1 = normal(loc=20,scale=5,size=300)  # First group: mean 20, SD 5, 300 points
sample2=normal(loc=40,scale=5,size=700)  # Second group: mean 40, SD 5, 700 points
sample = np.hstack((sample1,sample2))  # Combine both groups into one sample
plt.hist(sample,bins=50)  # Histogram may show two peaks because data is a mixture
```

```python
# REVISION: Create a Gaussian KDE model; bandwidth controls smoothness.
from sklearn.neighbors import KernelDensity  # KDE model from scikit-learn
model = KernelDensity(bandwidth=3,kernel="gaussian")  # Gaussian KDE; bandwidth controls smoothness
```

```python
# REVISION: Reshape data because sklearn expects 2-D input.
sample=sample.reshape((len(sample),1))  # sklearn expects 2-D input: rows=observations, columns=features
model.fit(sample)  # Learn the KDE from the sample
```

```python
# REVISION: Prepare x-values for KDE evaluation.
values = np.linspace(sample.min(),sample.max(),100)  # x-values where KDE will be evaluated
values = values.reshape((len(values),1))  # Convert x-values to sklearn's 2-D format
```

```python
# REVISION: score_samples returns log-density; exp() converts it to density.
probab = model.score_samples(values)  # Returns log-density, not ordinary density
prob = np.exp(probab)  # Convert log-density back to density
```

```python
# REVISION: Compare histogram density with the KDE curve.
plt.hist(sample,bins=50,density=True)  # Normalized histogram for comparison with KDE
plt.plot(values[:],prob)  # Plot the KDE density curve
plt.show()  # Display the plot
```

```python
# REVISION: Alternative Seaborn visualization of the distribution.
sns.distplot(sample,bins=50)  # Seaborn density visualization of the sample
```
