```python
import numpy as np
import pandas as pd , matplotlib.pyplot as plt , seaborn as sns
```

```python
# Flower data From seaborn website
Data = sns.load_dataset("iris")
Data.head(5) # check Data
Data["species"].unique() # check Unique Species
```

### Diffrentiating graph on the Basis of Length of petals and Sepal

```python
sns.kdeplot(Data,x="sepal_width",hue="species")
```

```python
sns.kdeplot(Data,x="sepal_length",hue="species")
```

```python
sns.kdeplot(Data,x="petal_width",hue="species")
```

```python
sns.kdeplot(Data,x="petal_length",hue="species")
sns.ecdfplot(Data,x="petal_length",hue="species")
```

### Observation
- here we can see we can diffrentiate the species of the flower on the basis of petal length and weidth
- two petal graph is better and remove other
- kdeplot gives the PDF 

```python
sns.kdeplot(Data,x="petal_width",hue="species")
sns.ecdfplot(Data,x="petal_width",hue="species")
```
