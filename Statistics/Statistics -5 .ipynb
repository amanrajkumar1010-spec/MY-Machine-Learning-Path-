## 3. Mathematical Principle: Area Under the Curve = 1

Continuous probability distributions ka universal rule:

> **Probability Density Function (PDF) ke curve ke neeche total area = 1**

Uniform distribution ka density plot rectangular hota hai.

For a Uniform Distribution:

* **Range:** `[a, b]`
* **Base:** `b - a`
* **Height / Probability Density:** `1 / (b - a)`
* **Area:**
  `(b - a) × 1/(b - a) = 1`

### Example: Uniform(0, 1)

* Base = `1 - 0 = 1`
* Height = `1 / (1 - 0) = 1`
* Area = `1 × 1 = 1`

## 4. Effect of Changing `high=1` to `high=2`

Suppose:

```python
x = np.random.uniform(low=0, high=1, size=1000)
```

is changed to:

```python
x = np.random.uniform(low=0, high=2, size=1000)
```

| Parameter                      | `Uniform(0,1)` | `Uniform(0,2)` |
| ------------------------------ | -------------: | -------------: |
| **Interval / Range**           |       `0 to 1` |       `0 to 2` |
| **Base** `(b-a)`               |            `1` |            `2` |
| **Height / Density** `1/(b-a)` |            `1` |          `0.5` |
| **Area**                       |    `1 × 1 = 1` |  `2 × 0.5 = 1` |

### Visual Takeaway

Jab range `0` se `1` ki jagah `0` se `2` ho jati hai, toh x-axis par distribution **2× wider** ho jata hai.

Total probability ko `1` hi rehna hota hai. Isliye density ki height:

* `Uniform(0,1)` → `1`
* `Uniform(0,2)` → `0.5`

Yaani range double hone par **probability density half** ho jati hai.

### Important Concept

**PDF ki height khud probability nahi hoti.**

Probability kisi interval ke liye **curve ke neeche ke area** se milti hai:

```text
Probability = Area under PDF curve
```

Isliye:

```text
Width ↑  →  Height ↓
```

taaki:

```text
Total Area = 1
```

hamesha maintain rahe.

## 5. `scipy.stats` se Uniform Distribution Fit Karna

### 1. `from scipy import stats`

```python
from scipy import stats
```

* SciPy ek **scientific computing library** hai.
* `stats` SciPy ka **statistics module** hai.
* Isme probability distributions, hypothesis testing, parameter estimation, PDF, CDF, etc. ke tools available hote hain.

---

### 2. `fit_params = stats.uniform.fit(x)`

```python
fit_params = stats.uniform.fit(x)
```

* `stats.uniform` SciPy mein **Uniform Distribution** ko represent karta hai.
* `.fit(x)` given data `x` ke basis par distribution ke **best-fit parameters** estimate karta hai.
* Conceptually, yeh **Maximum Likelihood Estimation (MLE)** ke through parameters estimate karta hai.
* Yeh normally ek tuple return karta hai:

```python
(loc, scale)
```

#### `loc`

* `loc` distribution ka **lower bound** hota hai.
* Isko hum `a` se represent kar sakte hain.
* Agar data `0` se generate hua hai, toh `loc` approximately `0` hoga.

#### `scale`

* `scale` distribution ki **width / total length** hoti hai.
* Formula:

```text
scale = b - a
```

* Agar distribution `0` se `2` tak hai:

```text
loc = 0
scale = 2 - 0 = 2
```

#### Upper Bound

SciPy mein:

```text
Upper Bound = loc + scale
```

Isliye:

```text
b = loc + scale
```

### Example

Agar:

```python
fit_params = (0, 2)
```

toh:

```text
loc   = 0
scale = 2
```

Aur distribution ka range hoga:

```text
0 → 2
```

---

### 3. `distribution = stats.uniform(loc=fit_params[0], scale=fit_params[1])`

```python
distribution = stats.uniform(
    loc=fit_params[0],
    scale=fit_params[1]
)
```

Yahan `fit_params` se estimated parameters ko `stats.uniform` mein pass kiya ja raha hai.

```python
fit_params[0]  →  loc
fit_params[1]  →  scale
```

Agar:

```python
fit_params = (0, 2)
```

toh internally yeh approximately equivalent hai:

```python
distribution = stats.uniform(loc=0, scale=2)
```

Ab `distribution` ek **frozen distribution object** hai.

### Frozen Distribution ka Meaning

Frozen distribution ka matlab hai ki distribution ke parameters ab **fixed** hain.

For example:

```python
distribution = stats.uniform(loc=0, scale=2)
```

ka matlab hai:

```text
Uniform(0, 2)
```

Ab isi object par different statistical calculations kar sakte hain:

```python
distribution.pdf(x)
distribution.cdf(x)
distribution.rvs()
```

* `.pdf(x)` → kisi point par **probability density** batata hai.
* `.cdf(x)` → `x` tak ki **cumulative probability** batata hai.
* `.rvs()` → fitted distribution se **random values generate** karta hai.

---

## 4. Complete Flow

Overall process ko is tarah samjho:

```python
from scipy import stats

fit_params = stats.uniform.fit(x)

distribution = stats.uniform(
    loc=fit_params[0],
    scale=fit_params[1]
)
```

### Step-by-Step

```text
Raw Data x
    ↓
stats.uniform.fit(x)
    ↓
(loc, scale)
    ↓
stats.uniform(loc=loc, scale=scale)
    ↓
Frozen Uniform Distribution
    ↓
PDF / CDF / Random Sampling
```

### Important Relationship

Uniform distribution ke liye:

```text
Lower Bound = loc
Width       = scale
Upper Bound = loc + scale
```

Aur PDF ki height:

```text
Height = 1 / scale
```

For example, agar:

```text
loc = 0
scale = 2
```

toh:

```text
Range  = 0 → 2
Width  = 2
Height = 1/2 = 0.5
Area   = 2 × 0.5 = 1
```

**Main idea:** `fit()` data se distribution ke parameters estimate karta hai, aur `stats.uniform(...)` un parameters ko use karke ek fixed/frozen theoretical distribution object banata hai.

# 5. Q-Q Plot ka Main Idea

```text
X-axis → Theoretical Quantiles
Y-axis → Sample Quantiles
```

Agar sample theoretical distribution ko closely follow karta hai, toh points approximately straight line ke around appear karte hain.

Conceptually:

```text
Sample Quantiles
      ↑
      |
      |        •
      |      •
      |    •
      |  •
      |•
      +----------------→ Theoretical Quantiles
```

---

# 6. Q-Q Plot ki Interpretation

### Points approximately straight line par

Agar points reference line ke around closely lie karte hain:

```text
Sample ≈ Theoretical Distribution
```

Toh sample data specified theoretical distribution ke saath **reasonably consistent** hai.

---

### Points line se systematically deviate karte hain

Agar points line se noticeable/systematic pattern mein deviate karte hain:

```text
Sample ≠ Theoretical Distribution
```

Toh data specified theoretical distribution ko achhi tarah follow nahi kar raha ho sakta hai.

**Important:** Sirf line se deviation dekhkar directly "skewness" ya "heavy tails" conclude nahi karna chahiye. Deviation ka pattern dekhna zaroori hai.

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# =====================================================================
# PART 1: Uniform Distribution (Line-by-Line Explanation)
# =====================================================================

# 1. Sample Data (20 se 80 ke beech uniform numbers)
np.random.seed(42) 
""" 1. Aapke aur mere screen par exact same numbers aayenge
Agar aap seed nahi lagate, to aapke PC par data generate hoga jaise [20.12, 79.88], aur mere PC par [20.01, 79.95].

Lekin jab hum np.random.seed(42) likhte hain, to:

Aapka output: Estimated Loc: 20.03, Estimated Scale: 59.94

Mera output: Estimated Loc: 20.03, Estimated Scale: 59.94"""
x = np.random.uniform(low=20, high=80, size=1000)

# MLE fit: fit_params[0] = loc (min value), fit_params[1] = scale (max - min)
fit_params = stats.uniform.fit(x) 

# Frozen object: Fit parameters assign karke continuous model tayar kiya
distribution = stats.uniform(loc=fit_params[0], scale=fit_params[1]) 

print("--- UNIFORM DISTRIBUTION ---")
print(f"Estimated Loc (Min):   {fit_params[0]:.2f}")
print(f"Estimated Scale (Width): {fit_params[1]:.2f}\n")


````

## Google Example

````python
# =====================================================================
# PART 2: Exponential Distribution (Waiting Time Example)
# =====================================================================

# 1. Sample Data: Customer support waiting time (True mean = 25 seconds)
# loc = 0 (wait time can't be negative), scale = 25
wait_times = np.random.exponential(scale=25, size=1500)

# 2. Fit karna:
# stats.expon.fit() MLE se loc aur scale dhoondhta hai.
# floc=0 se lower bound ko 0 par lock kiya (kyunki time >= 0 hota hai).
expon_params = stats.expon.fit(wait_times, floc=0)
loc_hat, scale_hat = expon_params

# 3. Fitted Distribution create karna:
fitted_expon = stats.expon(loc=loc_hat, scale=scale_hat)

print("--- EXPONENTIAL DISTRIBUTION ---")
print(f"Estimated Mean Time (Scale): {scale_hat:.2f} seconds")

# 4. Fitted model se prediction (e.g., 30 sec se zyada wait karne ki probability):
# sf (Survival Function) = 1 - cdf
prob_over_30 = fitted_expon.sf(30)
print(f"Probability wait > 30s:      {prob_over_30 * 100:.2f}%\n")


# =====================================================================
# PART 3: Data vs Fitted Curve Plot (Visual Proof)
# =====================================================================

plt.figure(figsize=(10, 5))

# Actual data ka normalized histogram
count, bins, _ = plt.hist(wait_times, bins=40, density=True, alpha=0.6, color='skyblue', edgecolor='black', label='Actual Wait Times')

# Fitted PDF line
x_axis = np.linspace(0, max(wait_times), 300)
fitted_pdf = fitted_expon.pdf(x_axis)
plt.plot(x_axis, fitted_pdf, 'r-', linewidth=2, label=f'Fitted PDF (scale={scale_hat:.1f})')

plt.title("Exponential Fit on Call Wait Times")
plt.xlabel("Wait Time (seconds)")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()
````
