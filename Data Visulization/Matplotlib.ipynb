# 📌 Matplotlib Quick Revision Summary

### 1. Data Setup & Imports

- **Libraries:** `matplotlib.pyplot` ko `plt` ki tarah aur `numpy` ko `np` ki tarah import kiya jata hai.
- **Array Creation:** `np.linspace(start, stop, num)` se linearly spaced points banaye gaye (jaise $x$ axis ke liye 0 se 5 tak 11 points).

---

### 2. Basic Line Plot

- **Code:** `plt.plot(x, y)`

- **Labels & Title:**
  - `plt.title("Title")`
  - `plt.xlabel("X-Axis")`
  - `plt.ylabel("Y-Axis")`

---

### 3. Subplots (`plt.subplot`)

- **Syntax:** `plt.subplot(rows, cols, index)`
- Ek hi figure me multiple graphs grid format me set karne ke liye use hota hai.
  - *Example (2x2 grid):* `plt.subplot(2, 2, 1)`, `plt.subplot(2, 2, 2)`, etc.

---

### 4. Object-Oriented Approach (OOA)

Canvas aur plots par pura control paane ke liye use hota hai:

- **Figure Creation:** `fig = plt.figure(figsize=(w, h), dpi=...)`
- **Add Axes:** `ax = fig.add_axes([left, bottom, width, height])`  
  *(values 0 se 1 ke beech fraction me hoti hain).*
- **Inset Plots:** Ek plot ke andar dusra plot place karne ke liye alag-alag coordinates ke sath multiple `add_axes` define kiye jate hain.
- **Reference Line:** `ax.axvline(x=val)` se vertical reference line draw hoti hai.

---

### 5. Common Plot Types

- **Line Plot:** `plt.plot(x, y)` — Continuous data/trends dekhne ke liye.
- **Scatter Plot:** `plt.scatter(x, y)` — Do variables ke beech correlation dekhne ke liye.
- **Histogram:** `plt.hist(data)` — Single variable ki frequency distribution dekhne ke liye (Y-axis automatically count calculate karti hai).
- **Box Plot:** `plt.boxplot(data)` — Outliers, median aur quartiles dekhne ke liye.

---

### 6. Image Handling

- **Import:** `import matplotlib.image as mpimg`
- **Load:** `img = mpimg.imread("path.png")`
- **Display:** `plt.imshow(img)`
- **Hide Axes:** `plt.axis('off')`
- **Cropping:** NumPy array slicing `img[y1:y2, x1:x2]` se image crop hoti hai.

---

````python
# Array Creation
x_axis = np.linspace(0,5,11) # 11 element 
x_axis
y_axis = x_axis**3
y_axis
````

### Plot creation

````python
plt.plot(x_axis,y_axis) #plot makes a line graph contain x and y as axis
plt.title("Graph")
plt.xlabel("x-axis   ------>")
plt.ylabel("y-axis   ------>")
````

### Sub PLots

````python
plt.subplot(2,2,1)
plt.plot(x_axis,y_axis)
plt.subplot(2,2,4)
plt.plot(x_axis,y_axis)
plt.subplot(2,2,2)
plt.plot(x_axis**2,x_axis**2)
plt.subplot(2,2,3)
plt.plot(y_axis,x_axis)
````

### Object Oriented Way
- Putting one graph on another
- Full controll over Graph

````python
fig = plt.figure()
axis1 = fig.add_axes([0.1,.01,0.5,1]) # percentage
axis1.plot(x_axis,y_axis)
axis2 = fig.add_axes([0.3,0.4,0.8,0.5])
axis2.plot(x_axis,y_axis)
````

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.

figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inch (pixel per inch).

````python
# trying to put on same graph
x = np.linspace(2,8,12)
y = x**2
y2 = x**3
fig = plt.figure(figsize=(12,8),dpi = 100)
axis3 = fig.add_axes([1,1,1,1])
axis3.plot(x,y,color = "Green")
axis3.plot(x,y2)

# What is axvline() for?

# axvline() creates a vertical line, for example: axis3.axvline(x=5) = verticle line at 5
````
