```python
import matplotlib.pyplot as plt
import seaborn as sns
import scipy
```

## Heat map

```python
flights = sns.load_dataset('flights')
flights
```

```python

tips = sns.load_dataset("tips")
tips
```

```python
tips_corr = tips[['total_bill','tip','size']]
```

```python
tips_corr
```

```python
corr_matrix = tips_corr.corr()
```

```python
sns.heatmap(corr_matrix,annot=True) # anote gives the value 
```

## Cluster Map

```python
sns.clustermap(corr_matrix)
#   - made by the use of hirarcy algo
```

## Pivot table Heat map

```python
pivflight = flights.pivot_table(values="passengers",index="month",columns='year')
pivflight
```

```python
sns.violinplot(data=pivflight)
plt.ylabel("Passanger")
```

```python
sns.heatmap(data=pivflight)
```
