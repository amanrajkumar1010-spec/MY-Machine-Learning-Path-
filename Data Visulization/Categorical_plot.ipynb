# Categorical Data Plots
### Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

- boxplot
- violinplot
- stripplot
- swarmplot
- barplot
- countplot
### Let's go through examples of each!

```python
file = sns.load_dataset('tips')
file
```

```python
sns.countplot(x = file['sex'],hue=file["smoker"])
plt.xlabel("Gender")   # defalut = Sex
plt.ylabel("No. of People") # default = Count
plt.show()
```

```python
sns.barplot(x = file["day"],y=file["total_bill"]) # take always 2 position (X,y)


# Graph say on an "average" what is the total bill on thu/daywise
```


You can change the estimator object to your own function, that converts a vector to a scalar Mean (default): np.mean Median: np.median Sum: np.sum Count: len (to count the number of observations) Standard Deviation: np.std Minimum: np.min Maximum: np.max

```python
import numpy as np
sns.barplot(x = file["sex"],y=file["total_bill"],estimator=np.sum)
```

### boxplot and violinplot

- boxplots and violinplots are used to shown the distribution of categorical data. 
- A box plot (or box-and-whisker plot) shows the distribution of quantitative data in a way that facilitates comparisons between variables or across levels of a categorical variable.
- The box shows the quartiles of the dataset while the whiskers extend to show the rest of the distribution, except for points that are determined to be “outliers” using a method that is a function of the inter-quartile range.   

```python
sns.violinplot(x="day",y = "total_bill",data=file,palette="rainbow")
plt.xlabel("Day Distribution")
plt.title("Vilone Plot")

# Observation
#  - sat have more veriance so its has more otliers 
```

# Stripplot vs. Swarmplot

### Stripplot
* **Type:** A scatterplot where one variable is categorical.
* **Usage:** Can be drawn standalone or as an overlay on a **box plot** or **violin plot** to display individual observations alongside aggregate distribution summaries.
* **Point Placement:** Points are plotted directly at their values, which can lead to overlapping (overplotting) when data density is high (though jitter can be applied).

---

### Swarmplot
* **Type:** A variation of a strip plot where points are positioned using a beeswarm algorithm.
* **Non-Overlapping:** Points are adjusted along the categorical axis so that they do not overlap, providing a clearer view of data distribution and density.
* **Scalability Limitations:**
  * **Visual Density:** Does not scale well to large datasets because points spread too wide and clutter the plot.
  * **Computational Cost:** Arranging non-overlapping points requires significantly more computation time as the number of data points increases.

---

### Quick Comparison

| Feature | `stripplot` | `swarmplot` |
| :--- | :--- | :--- |
| **Point Overlap** | Allows overlap (can add jitter) | Prevents overlap (adjusts position) |
| **Distribution Clarity** | Lower (due to overplotting) | Higher (resembles a distribution shape) |
| **Scalability (Large $N$)** | High (fast and handles large data) | Low (slow to compute, crowded display) |
| **Best Complements** | Box plots, Violin plots | Standalone or on top of Violin plots |

```python
sns.stripplot(x ="day",y="total_bill",data=file)
```
