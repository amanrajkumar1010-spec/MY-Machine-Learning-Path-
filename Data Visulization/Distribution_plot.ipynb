```python
import matplotlib.pyplot as plt
import seaborn as sns , numpy as nm, pandas as pd
```


### We use distribution plots to visualize the distribution of quantitative data.

Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

* distplot/histplot
* jointplot
* pairplot
* rugplot

### How to get Data
- seabon itslef have some pre loaded data on its git hub 
- so, You can download it and run through pandas
- or just name the file in sns.load_dataset

```python
D_load = sns.load_dataset("tips")
D_load
```

```python
D_load["size"].unique()  # This proves that size is also a catrogrical/qualitative data
```

### Creates Histogram
    - Distribution plot for Stats

```python
plt.subplot(1,2,1)
sns.histplot(D_load["total_bill"],bins =20,kde = True) # creates Histogram 
    # bin size are variable and contains Required No. dats for growth
    # bell curve(Normal Distribution) use
    # kernel-Density- estimettor 
            # use to know the squiness of the graphical_data using Veriance
plt.subplot(1,2,2)
sns.histplot(D_load["tip"],bins =20,kde = True)
```

### Joint Plot

```python
sns.jointplot(x = 'total_bill',y= 'tip',data = D_load,kind="reg")
# point at pair of x-axis and Y-axis
# this graph how: this cluster of dots at "tips"(2,4) and "total_bill"(10,30) so there are the most hightest number of values

# use in Linear regration
```

### Pair Plot

```python
sns.pairplot(D_load,kind= ("scatter"),hue="size",palette="rainbow") # Hue gives tells about the Distribution based on Gender
    #Scatter is use to find the corelation and co verince,co relation
```

### Rug plot

```python
sns.rugplot(D_load["total_bill"])
```
