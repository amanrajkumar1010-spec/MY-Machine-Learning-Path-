# Pandas Series

A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

## Creating Series
There are multiple ways to create a pandas Series:

* Here if List is vertical it is called Series.

* if it is Horizontal, it will be Arrays only that was used in Numpy aslo.

```python
import pandas as pd 
import numpy as np
```

```python
labels = ['a','b','c']
my_list = [10,20,30]
arr = np.array([50,60,70])
d = {1:100,2:200,3:300}        
```

```python
pd.Series(my_list)
pd.Series(my_list,) # Last wala hi work krta h 
pd.Series(data=my_list,index = labels)
```
