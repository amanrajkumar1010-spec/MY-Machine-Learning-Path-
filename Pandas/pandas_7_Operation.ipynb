 DataFrame Basic Operations: Quick Revision Notes

1. **`df.shape`**: DataFrame ke dimensions return karta hai tuple format me `(rows, columns)`[cite: 5].
2. **`df.columns`**: DataFrame ke sabhi column names ki list/Index show karta hai[cite: 5].
3. **`df.info()`**: DataFrame ka structural overview deta hai (row count, non-null counts, data types, and memory usage)[cite: 5].
4. **`df.describe()`**: Numerical columns ki descriptive statistics calculate karta hai (mean, std, min, percentiles [25%, 50%, 75%], max)[cite: 5].
5. **Vectorized Operations (`df['A'] * 10`)**: Loop lagaye bina poore column ke elements ko directly scalar value se transform ya multiply karta hai[cite: 5].
6. **`df['col'].apply(custom_func)`**: Ek custom defined function ko specific column ke har element par apply karta hai[cite: 5].
7. **In-place Persistence**: Function apply karne par changes tab tak DataFrame me save nahi hote jab tak use wapas column me assign na kiya jaye (`df['B'] = df['B'].apply(...)`)[cite: 5].
8. **`apply()` with `lambda`**: One-liner functions ke liye `apply(lambda x: x * 3)` ka use karke existing data modify ya naya column create kiya jata hai[cite: 5].

DataFrame Appling Function

```python
def square(x):
    return x**2
```

```python
df1["B"]= df1["B"].apply(square) # apply is the function
```

```python
df1["B"] # not saved by appling function only , get function saved in df1["B"] again
```

```python
df1["3B"] = df1["B"].apply(lambda x:x*3) # creaking new Table
df1
```
