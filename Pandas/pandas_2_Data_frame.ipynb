```python
import numpy as np
import pandas as pd
```

## Data Frame from Dict

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
d  = pd.Series(data=data)
print(d,sep ="\n") # sep does not work in pandas data frame 

```

```python
df = pd.DataFrame(data=data)
print(df)
```

```python
# If you specifically want TAB-separated output
# Use:

print(df.to_csv(sep="\t", index=False))
```

```python
print(df.to_string(index=False)) # alingment from left
```

## Data Frame from List

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2 = pd.DataFrame(data_list)
columns = ["Name","Age","City","Salary"]
df2 = pd.DataFrame(data_list,columns=columns,index= ["a","b","c","d"]) # horizontal(column name) is manual # index is also manual
df2
```

# Column Selecting 

```python
df2["Name"] # for targeting we should give column name (no index)
df2[["Name","City"]] # using Lsit of List
```

```python
# adding column
df2["Designation"] = ["Doctor","Eng","Teacher","chor"] # must be same no. element in list
df2   # Important: if we given new name to column to change previous name its does not chnage and make another column 
```

* Remove Column

```python
df2.drop("Disignation",axis= 1)
```

```python
df2   # Diginnation does not remove from main data frame 
# to remove from main data use key= "inplace"=True
```

```python
df2.drop("Disignation",axis =1,inplace=True)
df2  # once you remove any column and twice run the code it will not work as it is already deleted and no such column name exist 
```

* Multiple column remove

```python
df2.drop(["City","Salary"],axis=1)
```

Removing Rows

```python
df2.drop(["a","b"],axis=0) # also can use inplace to remove rom main data base , currenty only show how will it look after delition 
```
