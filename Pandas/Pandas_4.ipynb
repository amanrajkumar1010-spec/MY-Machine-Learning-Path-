        # Pandas DataFrame Combination: Quick Revision Notes

1. **`pd.merge(df1, df2, on='key', how='...')`**: Combines two DataFrames horizontally based on shared key columns.
2. **Inner Merge (`how='inner'`)**: Keeps only matching rows with keys present in both DataFrames.
3. **Outer Merge (`how='outer'`)**: Keeps all rows from both DataFrames, filling unmatched values with `NaN`.
4. **Left Merge (`how='left'`)**: Retains all rows from the left DataFrame and brings matching data from the right.
5. **Right Merge (`how='right'`)**: Retains all rows from the right DataFrame and brings matching data from the left.
6. **`pd.concat([df1, df2], axis=0)`**: Stacks DataFrames vertically (row-wise) by default.
7. **`pd.concat([df1, df2], axis=1)`**: Appends DataFrames horizontally (column-wise) along rows.
8. **`df1.join(df2, how='...')`**: Combines DataFrames primarily on their index labels (defaults to a left join).

```python
import numpy as np 
import pandas as pd 
```

* Merging 2 dataframe

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries, on='employee_id') # how='inner' is default , on is also default if not given
```

```python
pd.merge(employees,salaries, on='employee_id',how = 'outer') 
```

```python
pd.merge(employees,salaries,on='employee_id',how ='left')
```

```python
pd.merge(employees,salaries,on='employee_id',how ='right')
```

Concatination of 2 dataframes

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

```python
JODO = pd.concat([df1,df2])  # column wise
print(JODO)
```

```python
Hodo = pd.concat([df1,df2],axis = 1) # row wise
print(Hodo)
```

Joining 2 data freames


```python
df_1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df_2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

```python
df_1
```

```python
df_2
```

```python
df_1.join(df_2,how='outer')
```
