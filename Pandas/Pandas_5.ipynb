# GroupByAggregation

1. **List inputs**: Jab bhi multiple columns, rows ya dataFrames pass karne ho, unhe list format (jaise `[df1, df2]`) me pass karna hota hai[cite: 2].
2. **Default Merge Behaviour**: `pd.merge(df1, df2, on='employee_id')` me `how='inner'` default hota hai[cite: 2].
3. **Outer vs Inner Merge**: Outer merge missing data par `NaN` introduce karta hai, jabki inner merge unmatched rows ko drop kar deta hai[cite: 2].
4. **Vertical Concatenation (`axis=0`)**: By default bina axis diye concat karne par data row-wise judta hai[cite: 2].
5. **Horizontal Concatenation (`axis=1`)**: `axis=1` pass karne par data side-by-side column-wise concatenate hota hai[cite: 2].
6. **Index Joining**: `df_1.join(df_2)` direct index matching ke through data merge karta hai[cite: 2].
7. **Suffixes in Join**: Jab do DataFrames me same column names ho, tab clash avoid karne ke liye `lsuffix` aur `rsuffix` specify karna padta hai[cite: 2].

```python
import numpy as np 
import pandas as pd 
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
```

```python
df
```

Group by Category and calculate the sum of Sales

```python
df[df['Category'] == "A"]["Sales"].sum()
df.loc[df['Category'] == 'A', 'Sales'].sum()
df.groupby('Category')['Sales'].sum()
```

# Group by Store and calculate the sum of Sales

```python
df.groupby('Store')['Sales'].sum()
```

```python
cat = df.groupby('Category')
cat
for i,v in cat:
    print(i)
    print(v)
    print(type(v))
    print(v['Sales'].sum())
    
```

```python
# Group by multiple columns
# Group by Category and Store
cat = df.groupby(['Category','Store'])['Sales'].sum()
cat
```

Aggrigation
  - Basic function: Mean, Median,max,min,count,std etc

```python
Mean = df['Sales'].mean()
print(int(Mean))
```

```python
df['Sales'].agg(['sum', 'mean', 'min', 'max', 'count', 'std', 'median'])
```
