# Pandas Date Operations


## 1. Month Extract Karna

```python
df['Month'] = df['Date'].dt.month_name()
```

* `Date` column se **month ka naam** nikalta hai.
* `.dt` → datetime operations ke liye.
* `.month_name()` → month ka naam return karta hai.

**Example:**

`2025-05-20 → May`

---

## 2. Quarter Extract Karna

```python
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
```

* `.quarter` → quarter number nikalta hai.
* `.astype(str)` → number ko string mein convert karta hai.
* `'Q' +` → `Q1`, `Q2`, `Q3`, `Q4` format banata hai.

**Quarter Mapping:**

| Months    | Quarter |
| --------- | ------- |
| Jan – Mar | Q1      |
| Apr – Jun | Q2      |
| Jul – Sep | Q3      |
| Oct – Dec | Q4      |

---

## 3. Date ko Datetime mein Convert Karna

```python
df['Date'] = pd.to_datetime(df['Date'])
```

* `Date` column ko **datetime format** mein convert karta hai.
* `.dt` functions use karne ke liye Date ka datetime format mein hona zaroori hai.

---

## Quick Summary

* `pd.DataFrame()` → data ko table mein convert karta hai.
* `.dt` → date/time properties access karta hai.
* `.month_name()` → month ka naam deta hai.
* `.quarter` → quarter number deta hai.
* `.astype(str)` → datatype ko string mein convert karta hai.
* `pd.to_datetime()` → column ko datetime format mein convert karta hai.


````python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)
df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
df
````

Pivot Table

````python
pd.pivot_table(df,values = "Sales",index = 'Region',columns="Product",aggfunc = 'median')
````

````python
pd.pivot_table(df,values="Sales",columns="Product")  # default aggfunc is Mean
````

````python
pd.pivot_table(df,values=["Sales","Units"],index ="Region",columns="Product") # values will always be Numerical Value
````
