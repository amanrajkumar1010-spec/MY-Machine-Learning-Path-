# Finding Missing Data
        - Search = isna = row wise
        - remove = dropna = row wise
        - fill = fillna = coulmn wise
        - selection = column wise
        - column wise mean
        - Make it work in Main data frame = inplace= True

```python
import numpy as np 
import pandas as pd 


data = {
    'A': [1, 2, np.nan, 4, 5], # not a number value(None)
    'Ram': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
print(df)
```

```python
df.isna().sum() # null value to True # find total no. of null row wise
```

```python
df.isna().any() # does it have any row which have null value
```

- Remove Null value or missing Data

```python
df.dropna() # remove all row which have null value
```

```python
df
```

```python
df.dropna(thresh=3) # At least have 3 non - null values in rows
```

### Filling all the missing Value


```python
df.fillna(0) # fill na = column wise selection
```

```python
values = {"A":0,"Ram":100,"C":200,"D":400} # column wise 
df.fillna(values)
```

- Fill with Mean on missing value

```python
df
```

```python
df.fillna(df.mean()) # column wise mean
```
