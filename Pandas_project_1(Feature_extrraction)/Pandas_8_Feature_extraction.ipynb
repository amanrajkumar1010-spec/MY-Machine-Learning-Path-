```python
# Loading the Data
import pandas as pd
import numpy as np
df = pd.read_csv(r'anime.csv')
df
```

```python
df.loc[3]['Title'] #4th row see 
```

```python
def Extract_episode(txt):
    check = False
    data = ""
    for i in txt:
        if i=="(":
            check = True
        if check == True:
            data += i
        if i==")":
            check = False

    return int(data.replace("(","").replace("eps)",""))
print(Extract_episode('Gintama°TV (51 eps)Apr 2015 - Mar 2016605,113 members'))
```

```python
df["Episodes"] = df["Title"].apply(Extract_episode)
# df2.drop("Disignation",axis =1,inplace=True)
df
```

- Make a new column for time stamp

```python
df.loc[0]["Title"]
```

```python
def extraction_time(txt):
    check = False
    T_given = ""
    count = 0

    for i in txt:
        if i == ")":
            check = True
            continue

        if check and count < 19:
            T_given += i
            count += 1

        if count == 19:
            break

    return T_given

print(extraction_time(
    'Gintama°TV (51 eps)Apr 2015 - Mar 2016605,113 members'
))
```

```python
df["Time_stamp"] = df["Title"].apply(extraction_time)
df
```

#Extract the Months

```python
# Start Month nikalna (e.g., 'Apr')
df['start_month'] = df['Time_stamp'].str.extract(r'^([A-Za-z]+)')

# End Month nikalna (e.g., 'Jul')
df['end_month'] = df['Time_stamp'].str.extract(r'-\s*([A-Za-z]+)')
df
```

### Total Months

```python


# 1. Start aur End date strings ko alag karein
df[['start_str', 'end_str']] = df['Time_stamp'].str.split(' - ', expand=True)

# 2. Dono ko datetime me convert karein
df['start_date'] = pd.to_datetime(df['start_str'].str.strip(), format='%b %Y')
df['end_date'] = pd.to_datetime(df['end_str'].str.strip(), format='%b %Y')

# 3. Total months calculate karein
# Formula: (Year Difference * 12) + Month Difference
df['total_months'] = (df['end_date'].dt.year - df['start_date'].dt.year) * 12 + (df['end_date'].dt.month - df['start_date'].dt.month)

# Agar both months inclusive chahiye (jaise Apr 2009 se Jul 2010 = 16 months inclusive):
# df['total_months'] = df['total_months'] + 1

# Temporary columns drop karne ke liye:
df = df.drop(columns=['start_str', 'end_str'])
df
```
