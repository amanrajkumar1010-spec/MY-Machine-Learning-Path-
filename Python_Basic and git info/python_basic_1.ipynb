```python
a,b,c,d = 2,4,6,8.8
print(a,b,c,d)
print(a,b,c,d,sep="---",end=" ")
print(a,b,sep="---",end="\n")
name = "Aman"
print('%20s'% name)
num = 123.456789
num2 = 123456789
print('%2.3f'% num)
print('%4.3f'% num)
n1,n2,n3 = 10.365,20.4856,30.48568
print('first number is %2.3f, second number is %4.3f and third number is %5.3f'%(n1,n2,n3))
print('first number is {0:2.3f}, second number is {1:4.3f} and third number is {2:5.3f}'.format(n1,n2,n3))
print('first num is {n1}'.format(n1=n1))
print('first num is %d, second num is %i and third num is %s'%(n1,n2,n3))
```
